```python
from pathlib import Path
import pandas as pd
import json
from IPython.display import display, Image

# 项目根目录
PROJECT_ROOT = Path.cwd().resolve().parent

# 读取整合后的错句表
df = pd.read_csv(
    PROJECT_ROOT / "data/processed/errors_integrated_CJba.csv",
    dtype={"作文编码": "string"},
    encoding="utf-8-sig"
)

print(f"已加载 {len(df)} 条错句")
```

```python
def show_scans(composition_id):
    rows = df[df["作文编码"] == str(composition_id)]

    if rows.empty:
        print("未找到对应作文")
        return

    paths = json.loads(rows.iloc[0]["扫描图片路径"])

    print(f"作文编码：{composition_id}")
    print(f"关联错句：{len(rows)}条")
    print(f"扫描页数：{len(paths)}")

    for path in paths:
        display(Image(filename=str(PROJECT_ROOT / path)))
```

```python
show_scans("200307109523200252")
```
